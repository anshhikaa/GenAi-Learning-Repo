```python
### I) **Simple LLM call with streaming**
# 1) initialize a model 
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage 
llm = init_chat_model("groq:openai/gpt-oss-120b")
llm
#another way from langchain_groq import ChatGrok
# llm = ChatGrok("model: llama-3.1-8b-instant")  
# we use init_chat_model as it provides a unified interface to swap the model just change the string value
```

```python
# 2) create messages 
messages = ["SystemMessage : you are an helpful study assistent",
            "HumanMessage : what is the roadmap to study fastapi ?"
]
# 3 ) invoke model - return ai message
response = llm.invoke(messages)
response
print(response.content)
```

```python
# 4) streaming - return ai message chunk
for chunks in llm.stream(messages):
    print(chunks.content,end="",flush = True)
    
```

```python
### II)**Dynamic prompt template (how is it different from SystemMessage)** --> Translation app
from langchain_core.prompts import ChatPromptTemplate
dynamic_template = ChatPromptTemplate(
    [
        ("system" , "You are a professional translator. Translate the following text : {text} from {original_language} to {output_language} "),
        ("human" , "{text}")
    ]
)
final_prompt = dynamic_template.invoke(
    {
        "text" : "いってらっしゃい、エレン",
        "original_language" : "Japanese",
        "output_language" : "English"
    }
)
translated_response = llm.invoke(final_prompt)
print(translated_response.content)
```
